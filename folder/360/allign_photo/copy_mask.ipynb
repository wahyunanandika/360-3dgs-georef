```python
import os
import shutil

# path folder berisi gambar input
input_folder = r"E:\04.Pak_ival\Dataset\balai_kota_bandung\balai_kota\3.frame_360\3\GS033939"

# path file mask yang ingin dicopy
mask_path = r"E:\04.Pak_ival\Dataset\balai_kota_bandung\balai_kota\3.frame_360\mask_vide3.jpg"

# folder output
output_folder = r"E:\04.Pak_ival\Dataset\balai_kota_bandung\balai_kota\3.frame_360\mask_video3"
os.makedirs(output_folder, exist_ok=True)

# daftar file jpg dalam folder input
input_images = sorted([f for f in os.listdir(input_folder) if f.lower().endswith(".jpg")])

for img_name in input_images:
    # nama baru untuk mask
    base_name = os.path.splitext(img_name)[0]     # contoh: image_0001
    new_mask_name = f"{base_name}.png"

    # copy mask → output folder
    shutil.copy(mask_path, os.path.join(output_folder, new_mask_name))

print("Selesai! Mask telah dicopy mengikuti nama file pada input folder.")
```

```python
import os

folder_path = r"C:\dilans\360\cirebon\mask"

for fname in os.listdir(folder_path):
    if fname.endswith("_mask.png"):
        old_path = os.path.join(folder_path, fname)
        temp_name = fname.replace('_mask','_TEMP')
        temp_path = os.path.join(folder_path, temp_name)
        os.rename(old_path,temp_path)

for fname in os.listdir(folder_path):
    if fname.endswith("_TEMP.png"):
        old_path = os.path.join(folder_path, fname)
        final_name = fname.replace("_TEMP","")
        final_path = os.path.join(folder_path, final_name)
        os.rename(old_path, final_path)
```
