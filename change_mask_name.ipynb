```python
import os

folder_path = r"D:\dataset\aula_timur\LichtFeld_Export\masks"  # ganti sesuai path kamu

for filename in os.listdir(folder_path):
    old_path = os.path.join(folder_path, filename)

    if not os.path.isfile(old_path):
        continue

    new_name = filename

    # hapus _masked dulu (lebih panjang, biar nggak kepotong)
    if "_masked" in new_name:
        new_name = new_name.replace("_masked", "")
    # lalu hapus _mask
    if "_mask" in new_name:
        new_name = new_name.replace("_mask", "")

    if new_name != filename:
        new_path = os.path.join(folder_path, new_name)

        # safety check: jangan overwrite file existing
        if not os.path.exists(new_path):
            os.rename(old_path, new_path)
            print(f"Renamed: {filename} -> {new_name}")
        else:
            print(f"SKIP (already exists): {new_name}")
```
